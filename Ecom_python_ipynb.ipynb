In E-commerce, both the retailers (here brands) and the company have to make profit to sustain in the business. The E-Commerce company has the following rule for computing their own revenue:

The company charges

(i) 25% on the orders having final price (discounted price) greater than 600

(ii) 15% on the orders having final price (discounted price) greater than 350 but less than or equal to 600

(iii) 10% on the orders having final price (discounted price) greater than 100 but less than or equal to 350

(iv) Otherwise, 5% on the final price (discounted price)

```python
def compute_rev(x):
    if x > 600:
        return x*0.25
    elif 350 < x <= 600:
        return x*0.15
    elif 100 < x <=350 :
        return x*0.10
    else:
        return x*0.05

Ecom_dt['Revenue'] = Ecom_dt['discounted_price'].apply(compute_rev)
total_rev = Ecom_dt['Revenue'].sum()
print('The net revenue generated by the ECom company across all orders is', total_rev, 'rupees')
```

## Pairplot: Brand-Level Relationships

## Description

The pairplot compares four measures for each of the 2,484 brands: total retail price (RetailPriceBrandWise), total discounted price (DiscountPriceBrandWise), number of orders (No_Of_Orders) and brand revenue (BrandRevenueFinal). It draws a scatter plot for every pair of measures, and a histogram of each measure on the diagonal. This shows in one view how brand size, pricing and earnings move together.

## Observations

Prices move together almost perfectly. Total retail price and total discounted price have a correlation of about 0.95, so the points sit close to a straight upward line. A brand with higher retail value also has higher discounted value.
Brand revenue follows discounted price. Brand revenue is the discounted price minus the company’s commission, so the two are almost perfectly linear (correlation about 1.00). Brand revenue also tracks total retail price closely (about 0.95).
More orders mean more revenue. Number of orders and brand revenue have a strong positive correlation of about 0.80. The link between orders and prices is similar (about 0.83 with retail price and 0.79 with discounted price).
The diagonal histograms are heavily right-skewed. Most brands sit close to zero on every measure, and the median brand has just one order. About 60% of brands (1,491 of 2,484) have exactly one order, while the largest brand, ALLURE AUTO, has 469.
A few brands stand far from the rest. They are either brands with a very large number of orders, such as ALLURE AUTO (about ₹4.98 lakh brand revenue), or brands with a single very high-priced item, which stretches the price axes.

##Inference

Brand success in this marketplace comes mainly from volume rather than price. Brands with many orders earn the most brand revenue, and brand revenue rises in step with discounted sales value. The dense cluster of brands near zero and the few distant points show a long-tail market: a small number of brands generate most of the revenue, while thousands of brands have only one or a few orders. This supports the dashboard finding that the top 10 brands hold about 24% of brand revenue.
